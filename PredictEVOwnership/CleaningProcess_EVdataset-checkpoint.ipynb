```python
# Checking Datatypes, Mising Value, Unique Value:

def perform_basic_eda(data):
    temp = pd.DataFrame(index=data.columns)
    temp['data_type']=data.dtypes
    temp['null_count']=data.isnull().sum()
    temp['null_percentage']= round(data.isnull().sum()*100/len(df),2)
    temp['unique_count']=data.nunique()
    temp['unique_values'] = temp.index.map(lambda col: data[col].unique()[:10] if temp.loc[col, 'unique_count'] <= 10 else "Too many to display")
    return temp
data_info = perform_basic_eda(df)
```

```python
# Check how many duplicates are in the dataset
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
```

```python
# Look at the columns with more than 50% missing values
cols = df.columns[df.isnull().sum() * 100 / len(df) > 50.00]
print(f"Number of columns with more than 50% missing values are: {len(cols)}, {cols}")
```

```python
# Drop columns with more than 50% missing values
df.drop(cols, axis=1, inplace=True)
```

```python
# Identify columns with only one unique value and drop them
unique_counts = df.nunique()
single_value_columns = unique_counts[unique_counts == 1].index
df.drop(columns=single_value_columns, axis=1, inplace=True)
```

```python
# Remove all Accuracy columns because describe the confidence level and wont improve prediction
df = df[[col for col in df.columns if "accuracy" not in col.lower()]]
```

```python
# Remove Redundant columns:
redun_cols = ['premise_desc','flag_size','geo_area_text']
df.drop(columns=redun_cols, axis=1, inplace=True)
```

```python
# Look at the Shape
df.shape
```

```python
# Convert 'move_in_date' and 'home_bought_date' to datetime64[ns]:
df['move_in_date'] = pd.to_datetime(df['move_in_date'], errors='coerce')
df['home_bought_date'] = pd.to_datetime(df['home_bought_date'], format='%Y%m', errors='coerce')
```

```python
# Extract object columns again:
nonnumeric_columns = df.select_dtypes('object')
nonnumeric_columns.columns
```

```python
# Filling categorical columns:
for col in nonnumeric_columns:
    df[col].fillna('Unassigned', inplace=True)
```

```python
# Convert the datetime column to datetime format
df['home_bought_date'] = pd.to_datetime(df['home_bought_date'])

# Convert datetime to numerical format
df['date_numeric'] = (df['home_bought_date'] - pd.Timestamp('1970-01-01')) // pd.Timedelta('1D') # Unix epoch: standard starting point as a reference for time calculations

# Calculate the mean of the numerical values
mean_date_numeric = df['date_numeric'].mean()

# Convert the mean numerical value back to a datetime
mean_date = pd.Timestamp('1970-01-01') + pd.to_timedelta(mean_date_numeric, unit='D')

# Impute missing values with the mean date
df['home_bought_date'].fillna(mean_date, inplace=True)

# Drop the temporary numerical column
df.drop(columns=['date_numeric'], inplace=True)
```

```python
# Extract numeric columns:
numeric_columns = df.select_dtypes(exclude=['object','datetime64[ns]'])
print(f"Number of Numeric columns: {len(numeric_columns.columns)}")
```

```python
# Get the percent of missing values in numeric columns
percent_null_counts = round(numeric_columns.isnull().sum() * 100 / len(numeric_columns), 2).sort_values(ascending=False)
numeric_columns_havenull = [col for col in numeric_columns if df[col].isnull().any()]
df_numeric_havenull = df[numeric_columns_havenull]

print(f"Number of Numeric columns which have null: {len(numeric_columns_havenull)}")
```

```python
# Look at the distribution:
cols=['est_square_footage','est_alternative_financing','last_yr_avg_kwh_usage','last_yr_kwh_usage','can_afford_ev']

plt.figure(figsize=(15,12))
for i,col in enumerate(cols):
    plt.subplot(3,2,i+1)
    sns.distplot(df[col],bins=20)
    plt.title(f'Distribution of column {col}',fontsize=12)
    plt.xticks(rotation=90)
plt.subplots_adjust(hspace=0.8, top=0.6)
plt.show()
```

```python
# Impute with Median
cols=['est_square_footage','est_alternative_financing','last_yr_avg_kwh_usage','last_yr_kwh_usage','can_afford_ev', 'est_alternative_financing']

median_imputer = SimpleImputer(strategy="median")
df[cols] = median_imputer.fit_transform(df[cols])
```

```python
# Use Mode Imputation for categorical-like numeric columns and convert colums from float to int:
float_columns = df.select_dtypes('float').columns

df[float_columns] = df[float_columns].replace([np.inf, -np.inf], np.nan)
df[float_columns] = df[float_columns].apply(lambda col: col.fillna(col.mode()[0]))
float_columns = [col for col in float_columns if col not in ['can_afford_ev', 'est_alternative_financing']]

# Convert to integer
df[float_columns] = df[float_columns].astype(int)
```

```python
# Use Mode Imputation for categorical-like numeric columns:
numeric_columns = df.select_dtypes(exclude=['object','datetime64[ns]']).columns

mode_imputer = SimpleImputer(strategy="most_frequent")
df[numeric_columns] = mode_imputer.fit_transform(df[numeric_columns])
```

```python
# Heatmap
df_encoded = df.copy()

# Convert categorical columns to numeric using label encoding
for col in df_encoded.select_dtypes(include=['object']).columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

# Now, generate the heatmap with numeric values
plt.figure(figsize=(25,25))
sns.heatmap(df_encoded.corr(), annot=True, linewidths=0.5, fmt='.2f', cmap='BuGn')

plt.title("Correlation Heatmap")
plt.show()
```
